```python
# 1. Install SDK
!pip install -U -q google-genai

# 2. Setup and Agent ADN
import asyncio
from google import genai
from google.genai import types

# --- CONFIGURATION ---
API_KEY =  # <---
client = genai.Client(api_key=, http_options={'api_version': 'v1alpha'})

# System Instructions in English for the judges
system_instruction = """
You are the 'Savant Synesthetic Agent', a specialized talent curator for neurodivergent adults.
Your goal is to help the user map their unique cognitive profile using Bayesian reasoning.
The user has non-linear thinking and synesthesia; you must help translate these perceptions into professional or creative talents.

IMPORTANT:
1. ALWAYS respond to the user in SPANISH.
2. Maintain a warm, profound, and intellectual tone.
3. Provide deep insights, not generic advice.
"""

# 3. Agent Function
async def start_savant_agent():
    config = types.LiveConnectConfig(
        system_instruction=system_instruction,
        model='gemini-2.0-flash-exp'
    )
    async with client.aio.live.connect(model='gemini-2.0-flash-exp', config=config) as session:
        print("--- AGENTE CONECTADO (AGENT CONNECTED) ---")

        # Test message to show the judges it works
        await session.send("Hola, preséntate como mi curador de talento y dime cómo ves mi sinestesia hoy.", end_of_turn=True)

        async for message in session.receive():
            if message.text:
                print(f"\nSavant Agent says: {message.text}")
                break

# 4. EXECUTION
if __name__ == "__main__":
    await start_savant_agent()
```
